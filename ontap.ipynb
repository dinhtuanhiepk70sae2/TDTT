```python
#BÀI GIẢNG TUẦN 9
def binary_search(arr, value):
    left, right= 0, len(arr) - 1
    while left<= right:
        mid = (left + right) // 2
        if arr[mid] == value:
            return mid
        elif arr[mid] < value:
            left= mid + 1
        else:
            right= mid - 1
    return-1
arr = list(map(int, input().split()))
value = int(input())
print(binary_search(arr, value))
```

```python
#W7A2
def dem_so_lan_xuat_hien(arr, x) -> int :
    lst = []
    for i in range(len(arr)) :
        if arr[i] == x :
            lst.append(arr[i])
    return len(lst)
arr = list(map(int, input().split()))
x = int(input())
print(dem_so_lan_xuat_hien(arr, x))
```

```python
#insertion sort 
def insertion_sort(arr) :
    for i in range(1, len(arr)) :
        key = arr[i] 
        j = i - 1
        while j >= 0 and arr[j] > key :
            arr[j + 1] = key
            j -= 1
        arr[j + 1] = key 
    return arr
arr = list(map(int, input().split()))
print(insertion_sort(arr))
```
